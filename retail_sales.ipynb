```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\bhuky\Downloads\retail_sales_dataset.csv")
df.head()

```

```python
df.shape
df.info()
df.isnull().sum()
```

```python
# Fill missing numeric values with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
df
```

```python
# Fill missing numeric values with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
df
```

```python
# Convert Date column to datetime (if exists)
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])

print("Data cleaning completed successfully!")
```

```python
# Basic descriptive statistics
df.describe()
```

```python
# Total Amount statistics
mean_sales = df['Total Amount'].mean()
median_sales = df['Total Amount'].median()
mode_sales = df['Total Amount'].mode()[0]
std_sales = df['Total Amount'].std()

print("Total Amount Statistics:")
print("Mean:", mean_sales)
print("Median:", median_sales)
print("Mode:", mode_sales)
print("Standard Deviation:", std_sales)
```

```python
# Group total amount by date
daily_sales = df.groupby('Date')['Total Amount'].sum()

# Plot sales trend
plt.figure(figsize=(10,5))
plt.plot(daily_sales)
plt.title("Total Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Amount")
plt.show()
```

```python

# Age vs Total Amount
plt.figure(figsize=(8,5))
sns.scatterplot(x='Age', y='Total Amount', data=df)
plt.title("Age vs Total Amount")
plt.xlabel("Age")
plt.ylabel("Total Amount")
plt.show()

# Quantity vs Total Amount
plt.figure(figsize=(8,5))
sns.scatterplot(x='Quantity', y='Total Amount', data=df)
plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.show()
```

```python
# Quantity vs Total Amount
plt.figure(figsize=(8,5))
sns.barplot(x='Quantity', y='Total Amount', data=df)
plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.show()

# Price per Unit vs Total Amount
plt.figure(figsize=(8,5))
sns.scatterplot(x='Price per Unit', y='Total Amount', data=df)
plt.title("Price per Unit vs Total Amount")
plt.xlabel("Price per Unit")
plt.ylabel("Total Amount")
plt.show()
```

```python
# Heatmap for numeric columns
plt.figure(figsize=(8,6))
sns.heatmap(df[['Age','Quantity','Price per Unit','Total Amount']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
### Key Insights
- Total Amount is highly influenced by Quantity purchased
- Age of customers affects purchasing patterns slightly
- Price per Unit does not strongly affect total sales
- Sales trend over time shows peaks on certain dates
```
